### kspace data points and Min and max values
The acquired [MR signal](https://mriquestions.com/data-for-k-space.html) is already in a Fourier-like format suitable for filling the k-space matrix.
The MR signal is detected in quadrature. Each digitized data point of the MR signal can be represented as a **complex number**, with real and imaginary components.<br>
Alternatively, **each data point can be defined as having a magnitude and phase**, computed by simple trigonometric relations.

### Plotting kspace complex-valued image (middle slice from 16 one and last coil signal only)
To get from k-space to an actual image, we use the inverse Fourier transform (inverse_fft2_shift_1slice). This process translates the frequency data from k-space into spatial information, giving us a clear image. After applying the inverse Fourier transform to our k-space data, what we get isn’t just a regular image but a complex-valued image. This translates to each pixel having both a real and imaginary component. <br>
https://medium.com/@nadavlevi/visualising-mri-data-in-python-fe010c4a1c54

```python
# get kspace from kspace for the 8th slice and the last coil 
kspace_8th_slice_last_coil = kspace_8th_slice[18,:,:]
```

```python
# get kspace from kspace for the 8th slice and the last coil and apply log and abs to plot
plt.imshow(np.log(np.abs(kspace_8th_slice_last_coil) + 1e-9))
```

```python
# Extract the magnitude of the image data.
magnitude_k_data = np.abs(kspace_8th_slice_last_coil)

# Extract the phase of the image data.
phase_k_data = np.angle(kspace_8th_slice_last_coil)

# Separate out the real component of the image data.
real_k_data = np.real(kspace_8th_slice_last_coil)

# Separate out the imaginary component of the image data.
imaginary_k_data = np.imag(kspace_8th_slice_last_coil)

fig = plt.figure(figsize=(8,16))
list_plot = [magnitude_k_data, phase_k_data, real_k_data, imaginary_k_data]
list_title = ["Magnitude", "Phase", "Real", "Imaginary"]
for i in range(len(list_plot)):
    plt.subplot(1, len(list_plot), i + 1)
    plt.gca().set_title(list_title[i])
    plt.imshow(list_plot[i], cmap="gray")
    #plt.title("Kspace signal from 8th slice and last coils")
```

#### Check data range for different components

```python
print(f"Magnitude range: [{np.abs(kspace_8th_slice_last_coil).min():.2e}, {np.abs(kspace_8th_slice_last_coil).max():.2e}]")
print(f"Phase range: [{np.angle(kspace_8th_slice_last_coil).min():.2f}, {np.angle(kspace_8th_slice_last_coil).max():.2f}]")
print(f"Real range: [{np.real(kspace_8th_slice_last_coil).min():.2e}, {np.real(kspace_8th_slice_last_coil).max():.2e}]")
print(f"Imaginary range: [{np.imag(kspace_8th_slice_last_coil).min():.2e}, {np.imag(kspace_8th_slice_last_coil).max():.2e}]")
```

#### Normalize each subplot independently

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Magnitude
im0 = axes[0].imshow(np.abs(kspace_8th_slice_last_coil), cmap='gray')
axes[0].set_title('Magnitude')
axes[0].axis('off')
plt.colorbar(im0, ax=axes[0])

# Phase - use vmin/vmax for phase range
im1 = axes[1].imshow(np.angle(kspace_8th_slice_last_coil), cmap='gray', vmin=-np.pi, vmax=np.pi)
axes[1].set_title('Phase')
axes[1].axis('off')
plt.colorbar(im1, ax=axes[1])

# Real part - auto-scale or set range
im2 = axes[2].imshow(np.real(kspace_8th_slice_last_coil), cmap='gray')
axes[2].set_title('Real')
axes[2].axis('off')
plt.colorbar(im2, ax=axes[2])

# Imaginary part - auto-scale or set range
im3 = axes[3].imshow(np.imag(kspace_8th_slice_last_coil), cmap='gray')
axes[3].set_title('Imaginary')
axes[3].axis('off')
plt.colorbar(im3, ax=axes[3])

plt.tight_layout()
plt.show()
```

#### Solution 3: Manually set intensity ranges

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Magnitude
axes[0].imshow(np.abs(kspace_8th_slice_last_coil), cmap='gray')
axes[0].set_title('Magnitude')

# Phase
axes[1].imshow(np.angle(kspace_8th_slice_last_coil), cmap='gray', vmin=-np.pi, vmax=np.pi)
axes[1].set_title('Phase')

# Real part - use percentile-based windowing
real_part = np.real(kspace_8th_slice_last_coil)
vmin_real = np.percentile(real_part, 1)
vmax_real = np.percentile(real_part, 99)
axes[2].imshow(real_part, cmap='gray', vmin=vmin_real, vmax=vmax_real)
axes[2].set_title('Real')

# Imaginary part
imag_part = np.imag(kspace_8th_slice_last_coil)
vmin_imag = np.percentile(imag_part, 1)
vmax_imag = np.percentile(imag_part, 99)
axes[3].imshow(imag_part, cmap='gray', vmin=vmin_imag, vmax=vmax_imag)
axes[3].set_title('Imaginary')

for ax in axes:
    ax.axis('off')
    
plt.tight_layout()
plt.show()
```

### Combining signal from the 20 coils on the 16 slices

```python
number_of_slices = volume_kspace.shape[0]
number_of_slices
```

```python
# Plot all 16 slices in a 4x4 grid (RSS magnitude)
# add origin='lower' to plot upside down
number_of_slices = volume_kspace.shape[0]
fig, axes = plt.subplots(4, 4, figsize=(12,12))
axes = axes.flatten()
for idx, ax in enumerate(axes):
    if idx >= number_of_slices:
        ax.axis('off')
        continue
    # compute RSS magnitude for slice idx
    ksp = volume_kspace[idx]
    img_coils = inverse_fft2_shift_1slice(ksp)
    magnitude_slice = rsos(img_coils, 0)
    ax.imshow(magnitude_slice, cmap='gray', origin='lower')
    ax.set_title(f'Slice {idx}', fontsize=9)
    ax.axis('off')
fig.suptitle('RSS magnitude for each slices', fontsize=14)
plt.tight_layout(rect=[0,0,1,0.96])

```
